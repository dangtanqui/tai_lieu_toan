Tích vector chấm yêu cầu kích thước của hai vectơ phải giống nhau.


Hãy triển khai phiên bản sản phẩm chấm của riêng chúng ta bên dưới:

**Sử dụng vòng lặp for**, triển khai hàm trả về tích chấm của hai vectơ. Hàm trả về các đầu vào đã cho $a$ và $b$:
$$ x = \sum_{i=0}^{n-1} a_i b_i $$
Giả sử cả `a` và `b` đều có hình dạng giống nhau.


```python
def my_dot(a, b): 
    """
   Compute the dot product of two vectors
 
    Args:
      a (ndarray (n,)):  input vector 
      b (ndarray (n,)):  input vector with same dimension as a
    
    Returns:
      x (scalar): 
    """
    x=0
    for i in range(a.shape[0]):
        x = x + a[i] * b[i]
    return x
```

```python
# kiểm tra 1-Da = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
print(f"my_dot(a, b) = {my_dot(a, b)}")
```

Lưu ý, tích số chấm dự kiến ​​sẽ trả về giá trị vô hướng. 

Hãy thử các thao tác tương tự bằng cách sử dụng `np.dot`.


```python
# kiểm tra 1-Da = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
c = np.dot(a, b)
print(f"NumPy 1-D np.dot(a, b) = {c}, np.dot(a, b).shape = {c.shape} ") 
c = np.dot(b, a)
print(f"NumPy 1-D np.dot(b, a) = {c}, np.dot(a, b).shape = {c.shape} ")
```

### 3.4.7 Nhu cầu về tốc độ: vector và vòng lặp for
chúng ta sử dụng thư viện NumPy vì nó cải thiện hiệu suất bộ nhớ tốc độ. Hãy chứng minh:


```python
np.random.seed(1)
a = np.random.rand(10000000)  # very large arrays
b = np.random.rand(10000000)

tic = time.time()  # capture start time
c = np.dot(a, b)
toc = time.time()  # capture end time

print(f"np.dot(a, b) =  {c:.4f}")
print(f"Vectorized version duration: {1000*(toc-tic):.4f} ms ")

tic = time.time()  # capture start time
c = my_dot(a,b)
toc = time.time()  # capture end time

print(f"my_dot(a, b) =  {c:.4f}")
print(f"loop version duration: {1000*(toc-tic):.4f} ms ")

del(a);del(b)  #remove these big arrays from memory
```

Vì vậy, vector hóa mang lại tốc độ lớn trong ví dụ này. Điều này là do NumPy tận dụng tốt hơn tính song song của dữ liệu có sẵn trong phần cứng cơ bản.
```python
a = np.zeros((1, 5))                                       
print(f"a shape = {a.shape}, a = {a}")                     

a = np.zeros((2, 1))                                                                   
print(f"a shape = {a.shape}, a = {a}") 

a = np.random.random_sample((1, 1))  
print(f"a shape = {a.shape}, a = {a}") 
```

Người ta cũng có thể chỉ định dữ liệu theo cách thủ công. Kích thước được chỉ định bằng các dấu ngoặc bổ sung phù hợp với định dạng in ở trên.


```python
# NumPy các quy trình phân bổ bộ nhớ và điền các giá trị do người dùng chỉ địnha = np.array([[5], [4], [3]]);   print(f" a shape = {a.shape}, np.array: a = {a}")
a = np.array([[5],   # One can also
              [4],   # separate values
              [3]]); #into separate rows
print(f" a shape = {a.shape}, np.array: a = {a}")
```

<a name="toc_40015_4.4.1"></a>
### 4.4.1 Lập chỉ mục


Ma trận bao gồm chỉ số thứ hai. Hai chỉ mục mô tả [hàng, cột]. Access có thể trả về một phần tử hoặc một hàng/cột. Xem bên dưới:


```python
# các phép toán lập chỉ mục vector trên ma trậna = np.arange(6).reshape(-1, 2)   #reshape is a convenient way to create matrices
print(f"a.shape: {a.shape}, \na= {a}")

# truy cập một phần tửprint(f"\na[2,0].shape:   {a[2, 0].shape}, a[2,0] = {a[2, 0]},     type(a[2,0]) = {type(a[2, 0])} Accessing an element returns a scalar\n")

# truy cập một hàngprint(f"a[2].shape:   {a[2].shape}, a[2]   = {a[2]}, type(a[2])   = {type(a[2])}")
```

Điều đáng chú ý là ví dụ cuối cùng. Việc truy cập ma trận chỉ bằng cách chỉ định hàng sẽ trả về *vectơ 1-D*.


**Định hình lại**  
Ví dụ trước sử dụng [reshape](https://numpy.org/doc/stable/reference/generated/numpy.reshape.html) để định hình mảng.  
`a = np.arange(6).reshape(-1, 2) `   
Dòng mã này lần đầu tiên tạo ra một *Vectơ 1-D* gồm sáu phần tử. Sau đó, nó định hình lại vectơ đó thành mảng *2-D* bằng lệnh định hình lại. Điều này có thể đã được viết:  
`a = np.arange(6).reshape(3, 2) `  
Để có cùng một mảng 3 hàng, 2 cột.
Đối số -1 yêu cầu quy trình tính toán số hàng dựa trên kích thước của mảng và số cột.


<a name="toc_40015_4.4.2"></a>
### 4.4.2 Cắt lát
Việc cắt lát tạo ra một mảng chỉ mục bằng cách sử dụng bộ ba giá trị (`start:stop:step`). Một tập hợp con các giá trị cũng hợp lệ. Việc sử dụng nó được giải thích rõ nhất bằng ví dụ:


```python
# hoạt động cắt vector 2-Da = np.arange(20).reshape(-1, 10)
print(f"a = \n{a}")

# truy cập 5 phần tử liên tiếp (bắt đầu: dừng: bước)print("a[0, 2:7:1] = ", a[0, 2:7:1], ",  a[0, 2:7:1].shape =", a[0, 2:7:1].shape, "a 1-D array")

# truy cập 5 phần tử liên tiếp (bắt đầu: dừng: bước) thành hai hàngprint("a[:, 2:7:1] = \n", a[:, 2:7:1], ",  a[:, 2:7:1].shape =", a[:, 2:7:1].shape, "a 2-D array")

# truy cập tất cả các yếu tốprint("a[:,:] = \n", a[:,:], ",  a[:,:].shape =", a[:,:].shape)

# truy cập tất cả các phần tử trong một hàng (cách sử dụng rất phổ biến)print("a[1,:] = ", a[1,:], ",  a[1,:].shape =", a[1,:].shape, "a 1-D array")
# giống nhưprint("a[1]   = ", a[1],   ",  a[1].shape   =", a[1].shape, "a 1-D array")
```
