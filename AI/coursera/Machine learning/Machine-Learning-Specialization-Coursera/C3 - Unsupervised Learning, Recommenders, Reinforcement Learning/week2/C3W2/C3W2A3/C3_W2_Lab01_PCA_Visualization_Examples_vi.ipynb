## Ví dụ về bài giảng

Chúng ta sẽ làm việc trên cùng một ví dụ mà Andrew đã trình bày trong bài giảng.


```python
X = np.array([[ 99,  -1],
       [ 98,  -1],
       [ 97,  -2],
       [101,   1],
       [102,   1],
       [103,   2]])
```

```python
plt.plot(X[:,0], X[:,1], 'ro')
```

```python
# Đang tải thuật toán PCA
pca_2 = PCA(n_components=2)
pca_2
```

```python
# Hãy fit dữ liệu. Chúng ta không cần phải mở rộng quy mô vì việc triển khai sklearn đã xử lý được nó.
pca_2.fit(X)
```

```python
pca_2.explained_variance_ratio_
```

Tọa độ trên component chính đầu tiên (trục đầu tiên) đủ để giữ lại 99,24% thông tin ("variance giải thích").  Component chính thứ hai bổ sung thêm 0,76% thông tin ("variance được giải thích") không được lưu trữ trong tọa độ component chính đầu tiên.


```python
X_trans_2 = pca_2.transform(X)
X_trans_2
```

Hãy coi cột 1 là tọa độ dọc theo trục chính thứ nhất component (trục mới đầu tiên) và cột 2 là tọa độ dọc theo trục chính thứ hai component (trục mới thứ hai).

Bạn có thể chỉ cần chọn component chính đầu tiên vì nó giữ lại 99% thông tin (variance giải thích).


```python
pca_1 = PCA(n_components=1)
pca_1
```

```python
pca_1.fit(X)
pca_1.explained_variance_ratio_
```

```python
X_trans_1 = pca_1.transform(X)
X_trans_1
```

Lưu ý cột này chỉ là cột đầu tiên của `X_trans_2`.


Nếu bạn có 2 features (hai cột dữ liệu) và chọn 2 components chính thì bạn sẽ giữ nguyên tất cả thông tin và dữ liệu sẽ giống như ban đầu.


```python
X_reduced_2 = pca_2.inverse_transform(X_trans_2)
X_reduced_2
```

```python
plt.plot(X_reduced_2[:,0], X_reduced_2[:,1], 'ro')
```

Giảm xuống 1 chiều thay vì 2


```python
X_reduced_1 = pca_1.inverse_transform(X_trans_1)
X_reduced_1
```

```python
plt.plot(X_reduced_1[:,0], X_reduced_1[:,1], 'ro')
```

Lưu ý rằng dữ liệu hiện chỉ nằm trên một dòng (dòng này là component chính duy nhất được sử dụng để mô tả dữ liệu và mỗi ví dụ có một "tọa độ" duy nhất dọc theo trục đó để mô tả vị trí của nó.

## Sử dụng PCA trong Phân tích dữ liệu thăm dò


Hãy tải một món đồ chơi dataset với các mẫu </sub> và </sub> features.


```python
df = pd.read_csv("toy_dataset.csv")
```

```python
df.head()
```

Đây là dataset có </sub> features.

Hãy thử xem liệu có mẫu nào trong dữ liệu không. Hàm sau sẽ lấy mẫu ngẫu nhiên 100 bộ dữ liệu theo cặp (x, y) của features, vì vậy chúng ta có thể scatter-plot chúng.


```python
def get_pairs(n = 100):
    from random import randint
    i = 0
    tuples = []
    while i < 100:
        x = df.columns[randint(0,999)]
        y = df.columns[randint(0,999)]
        while x == y or (x,y) in tuples or (y,x) in tuples:
            y = df.columns[randint(0,999)]
        tuples.append((x,y))
        i+=1
    return tuples
            
```

```python
pairs = get_pairs()
```

Bây giờ hãy plot chúng!


```python
fig, axs = plt.subplots(10,10, figsize = (35,35))
i = 0
for rows in axs:
    for ax in rows:
        ax.scatter(df[pairs[i][0]],df[pairs[i][1]], color = "#C00000")
        ax.set_xlabel(pairs[i][0])
        ax.set_ylabel(pairs[i][1])
        i+=1
```

Có vẻ như không có nhiều thông tin ẩn trong features theo cặp. Ngoài ra, không thể kiểm tra mọi kết hợp do số lượng features. Chúng ta hãy thử xem mối tương quan tuyến tính giữa chúng.


```python
# Quá trình này có thể mất 1 phút để chạy
corr = df.corr()
```

```python
# # Điều này sẽ hiển thị tất cả features có tương quan > 0,5 ở giá trị tuyệt đối. Chúng ta loại bỏ features
# # có tương quan == 1 để loại bỏ mối tương quan của feature với chính nó

mask = (abs(corr) > 0.5) & (abs(corr) != 1)
corr.where(mask).stack().sort_values()
```

Mối tương quan tối đa và tối thiểu nằm trong khoảng `0.631 - 0.632`. Điều này cũng không thể hiện quá nhiều.


Hãy thử phân tách PCA để nén dữ liệu của chúng ta thành không gian con (mặt phẳng) 2 chiều để chúng ta có thể plot dưới dạng scatter plot.


```python
# Đang tải đối tượng PCA
pca = PCA(n_components = 2) # Here we choose the number of components that we will keep.
X_pca = pca.fit_transform(df)
df_pca = pd.DataFrame(X_pca, columns = ['principal_component_1','principal_component_2'])
```

```python
df_pca.head()
```

```python
plt.scatter(df_pca['principal_component_1'],df_pca['principal_component_2'], color = "#C00000")
plt.xlabel('principal_component_1')
plt.ylabel('principal_component_2')
plt.title('PCA decomposition')
```

Điều này thật tuyệt vời! Chúng ta có thể thấy các cụm được xác định rõ ràng.


```python
# Pca.explained_variance_ration_ trả về một danh sách trong đó nó hiển thị lượng variance được giải thích bởi mỗi component chính.
sum(pca.explained_variance_ratio_)
```

Và chúng ta chỉ bảo toàn được khoảng 14,6% variance!


Khá ấn tượng! Chúng ta có thể thấy rõ các cụm trong dữ liệu của chúng ta, điều mà trước đây chúng ta không thể nhìn thấy. Bạn có thể nhận ra bao nhiêu cụm? 8, 10? 

Nếu chúng ta chạy kích thước PCA đến plot 3, chúng ta sẽ nhận được nhiều thông tin hơn từ dữ liệu.


```python
pca_3 = PCA(n_components = 3).fit(df)
X_t = pca_3.transform(df)
df_pca_3 = pd.DataFrame(X_t,columns = ['principal_component_1','principal_component_2','principal_component_3'])
```

```python
import plotly.express as px
```

```python
fig = px.scatter_3d(df_pca_3, x = 'principal_component_1', y = 'principal_component_2', z = 'principal_component_3').update_traces(marker = dict(color = "#C00000"))
fig.show()
```

```python
sum(pca_3.explained_variance_ratio_)
```
