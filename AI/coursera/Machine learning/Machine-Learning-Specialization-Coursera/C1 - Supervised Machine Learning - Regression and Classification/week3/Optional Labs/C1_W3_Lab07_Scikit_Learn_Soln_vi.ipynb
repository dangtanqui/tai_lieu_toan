# Lab chưa được phân loại: Logistic Regression sử dụng Scikit-Learn


## Mục tiêu
Trong lab này, bạn sẽ:
- Training logistic regression model bằng scikit-learn.


## Dataset 
Hãy bắt đầu với dataset giống như trước đây.


```python
import numpy as np

X = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1])
```

## Lắp model

Code bên dưới nhập [logistic regression model](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LogisticRegression.html#sklearn.linear_model.LogisticRegression) từ scikit-learn. Bạn có thể điều chỉnh model này trên dữ liệu training bằng cách gọi hàm `fit`.


```python
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression()
lr_model.fit(X, y)
```

## Tạo Predictions

Bạn có thể thấy predictions được tạo bởi model này bằng cách gọi hàm `predict`.


```python
y_pred = lr_model.predict(X)

print("Prediction on training set:", y_pred)
```

## Tính accuracy

Bạn có thể tính toán accuracy này của model này bằng cách gọi hàm `score`.


```python
print("Accuracy on training set:", lr_model.score(X, y))
```
