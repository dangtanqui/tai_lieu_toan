- Huấn luyện mô hình logistic regression bằng scikit-learn.


## Bộ dữ liệu 
Hãy bắt đầu với cùng một tập dữ liệu như trước.


```python
import numpy as np

X = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1])
```

## Phù hợp với mô hình

Mã bên dưới nhập [logistic regression model](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LogisticRegression.html#sklearn.linear_model.LogisticRegression) từ scikit-learn. Bạn có thể điều chỉnh mô hình này trên dữ liệu huấn luyện bằng cách gọi hàm `fit`.


```python
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression()
lr_model.fit(X, y)
```

## Đưa ra dự đoán

Bạn có thể xem các dự đoán do mô hình này đưa ra bằng cách gọi hàm `predict`.


```python
y_pred = lr_model.predict(X)

print("Prediction on training set:", y_pred)
```

##Tính toán độ chính xác

Bạn có thể tính toán độ chính xác này của mô hình này bằng cách gọi hàm `score`.


```python
print("Accuracy on training set:", lr_model.score(X, y))
```
