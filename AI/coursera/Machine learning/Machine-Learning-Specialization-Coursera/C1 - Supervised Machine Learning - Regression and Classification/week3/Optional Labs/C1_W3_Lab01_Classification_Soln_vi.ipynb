## Vấn đề phân loại
<img align="left" src="./images/C1_W3_Classification.png"     style=" width:380px; padding: 10px; " > Ví dụ về các vấn đề phân loại là những việc như: xác định email là Thư rác hay Không phải Thư rác hoặc xác định xem khối u là ác tính hay lành tính. Đặc biệt, đây là những ví dụ về phân loại *nhị phân* trong đó có hai kết quả có thể xảy ra.  Kết quả có thể được mô tả theo cặp 'tích cực'/'tiêu cực' chẳng hạn như 'có'/'không, 'đúng'/'sai' hoặc '1'/'0'. 

Các sơ đồ của tập dữ liệu phân loại thường sử dụng các ký hiệu để biểu thị kết quả của một ví dụ. Trong các biểu đồ bên dưới, 'X' được sử dụng để biểu thị các giá trị dương trong khi 'O' biểu thị các kết quả âm.


```python
x_train = np.array([0., 1, 2, 3, 4, 5])
y_train = np.array([0,  0, 0, 1, 1, 1])
X_train2 = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y_train2 = np.array([0, 0, 0, 1, 1, 1])
```

```python
pos = y_train == 1
neg = y_train == 0

fig,ax = plt.subplots(1,2,figsize=(8,3))
# âm mưu 1, biến đơn
ax[0].scatter(x_train[pos], y_train[pos], marker='x', s=80, c = 'red', label="y=1")
ax[0].scatter(x_train[neg], y_train[neg], marker='o', s=100, label="y=0", facecolors='none', 
              edgecolors=dlc["dlblue"],lw=3)

ax[0].set_ylim(-0.08,1.1)
ax[0].set_ylabel('y', fontsize=12)
ax[0].set_xlabel('x', fontsize=12)
ax[0].set_title('one variable plot')
ax[0].legend()

# âm mưu 2, hai biến
plot_data(X_train2, y_train2, ax[1])
ax[1].axis([0, 4, 0, 4])
ax[1].set_ylabel('$x_1$', fontsize=12)
ax[1].set_xlabel('$x_0$', fontsize=12)
ax[1].set_title('two variable plot')
ax[1].legend()
plt.tight_layout()
plt.show()
```

Lưu ý trong các ô trên:
- Trong biểu đồ biến đơn, kết quả dương tính được hiển thị cả chữ 'X' màu đỏ và y=1. Kết quả âm tính có chữ 'O' màu xanh lam và nằm ở y=0.
   - Nhớ lại trong trường hợp linear regression, y sẽ không bị giới hạn ở hai giá trị mà có thể là bất kỳ giá trị nào.
- Trong đồ thị hai biến không có trục y.  Kết quả dương tính được hiển thị dưới dạng chữ 'X' màu đỏ, trong khi kết quả âm tính sử dụng ký hiệu 'O' màu xanh lam.
    - Nhớ lại trong trường hợp linear regression có nhiều biến, y sẽ không bị giới hạn ở hai giá trị và một đồ thị tương tự sẽ là ba chiều.
