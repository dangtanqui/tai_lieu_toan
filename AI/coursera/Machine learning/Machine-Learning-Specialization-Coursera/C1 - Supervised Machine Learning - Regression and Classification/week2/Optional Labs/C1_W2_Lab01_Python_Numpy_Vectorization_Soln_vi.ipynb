### 3.4.6 Vector Vector dot product
dot product là trụ cột của Đại số tuyến tính và NumPy. Đây là một thao tác được sử dụng rộng rãi trong khóa học này và cần được hiểu rõ. dot product được hiển thị bên dưới.

Dot product nhân các giá trị trong hai phần tử vectors rồi tính tổng kết quả.
Vector dot product yêu cầu kích thước của hai vectors phải giống nhau.


Hãy triển khai phiên bản dot product của riêng chúng ta bên dưới:

**Sử dụng vòng lặp for**, triển khai hàm trả về dot product của hai vectors. Hàm trả về các đầu vào đã cho $a$ và $b$:
$$ x = \sum_{i=0}^{n-1} a_i b_i $$
Giả sử cả `a` và `b` đều là shape giống nhau.


```python
def my_dot(a, b): 
    """
   Compute the dot product of two vectors
 
    Args:
      a (ndarray (n,)):  input vector 
      b (ndarray (n,)):  input vector with same dimension as a
    
    Returns:
      x (scalar): 
    """
    x=0
    for i in range(a.shape[0]):
        x = x + a[i] * b[i]
    return x
```

```python
# kiểm tra 1-Da = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
print(f"my_dot(a, b) = {my_dot(a, b)}")
```

Lưu ý, dot product dự kiến ​​sẽ trả về giá trị scalar. 

Hãy thử các thao tác tương tự bằng `np.dot`.


```python
# kiểm tra 1-Da = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
c = np.dot(a, b)
print(f"NumPy 1-D np.dot(a, b) = {c}, np.dot(a, b).shape = {c.shape} ") 
c = np.dot(b, a)
print(f"NumPy 1-D np.dot(b, a) = {c}, np.dot(a, b).shape = {c.shape} ")
```

Ở trên, bạn sẽ lưu ý rằng kết quả của 1-D phù hợp với việc triển khai của chúng ta.

### 3.4.7 Nhu cầu về tốc độ: vector so với vòng lặp for
Chúng ta sử dụng thư viện NumPy vì nó cải thiện hiệu suất bộ nhớ tốc độ. Hãy chứng minh:


```python
np.random.seed(1)
a = np.random.rand(10000000)  # very large arrays
b = np.random.rand(10000000)

tic = time.time()  # capture start time
c = np.dot(a, b)
toc = time.time()  # capture end time

print(f"np.dot(a, b) =  {c:.4f}")
print(f"Vectorized version duration: {1000*(toc-tic):.4f} ms ")

tic = time.time()  # capture start time
c = my_dot(a,b)
toc = time.time()  # capture end time

print(f"my_dot(a, b) =  {c:.4f}")
print(f"loop version duration: {1000*(toc-tic):.4f} ms ")

del(a);del(b)  #remove these big arrays from memory
```

Vì vậy, vectorization cung cấp tốc độ tăng tốc lớn trong ví dụ này. Điều này là do NumPy tận dụng tốt hơn khả năng song song dữ liệu có sẵn trong phần cứng cơ bản. Hãy cùng khám phá một số thao tác sử dụng matrices.


<a name="toc_40015_4.4.1"></a>
### 4.4.1 Lập chỉ mục


Matrices bao gồm chỉ mục thứ hai. Hai chỉ mục mô tả [hàng, cột]. Access có thể trả về một phần tử hoặc một hàng/cột. Xem bên dưới:


```python
# Hoạt động lập chỉ mục vector trên matricesa = np.arange(6).reshape(-1, 2)   #reshape is a convenient way to create matrices
print(f"a.shape: {a.shape}, \na= {a}")

# truy cập một phần tửprint(f"\na[2,0].shape:   {a[2, 0].shape}, a[2,0] = {a[2, 0]},     type(a[2,0]) = {type(a[2, 0])} Accessing an element returns a scalar\n")

# truy cập một hàngprint(f"a[2].shape:   {a[2].shape}, a[2]   = {a[2]}, type(a[2])   = {type(a[2])}")
```

Điều đáng chú ý là ví dụ cuối cùng. Truy cập matrix chỉ bằng cách chỉ định hàng sẽ trả về *1-D vector*.


**Định hình lại**  
Ví dụ trước sử dụng [reshape](https://numpy.org/doc/stable/reference/generated/numpy.reshape.html) đến shape array.  
`a = np.arange(6).reshape(-1, 2) `   
Dòng code này lần đầu tiên tạo ra *1-D Vector* gồm sáu phần tử. Sau đó, nó định hình lại vector đó thành *2-D* array bằng lệnh định hình lại. Điều này có thể đã được viết:  
`a = np.arange(6).reshape(3, 2) `  
Để đến cùng 3 hàng, 2 cột array.
Đối số -1 yêu cầu quy trình tính toán số hàng dựa trên kích thước của array và số cột.


<a name="toc_40015_4.4.2"></a>
### 4.4.2 Cắt lát
Việc cắt lát tạo ra một array gồm các chỉ mục bằng cách sử dụng một bộ ba giá trị (`start:stop:step`). Một tập hợp con các giá trị cũng hợp lệ. Việc sử dụng nó được giải thích rõ nhất bằng ví dụ:


```python
# vector Hoạt động cắt 2-Da = np.arange(20).reshape(-1, 10)
print(f"a = \n{a}")

# truy cập 5 phần tử liên tiếp (bắt đầu: dừng: bước)print("a[0, 2:7:1] = ", a[0, 2:7:1], ",  a[0, 2:7:1].shape =", a[0, 2:7:1].shape, "a 1-D array")

# truy cập 5 phần tử liên tiếp (bắt đầu: dừng: bước) thành hai hàngprint("a[:, 2:7:1] = \n", a[:, 2:7:1], ",  a[:, 2:7:1].shape =", a[:, 2:7:1].shape, "a 2-D array")

# truy cập tất cả các yếu tốprint("a[:,:] = \n", a[:,:], ",  a[:,:].shape =", a[:,:].shape)

# truy cập tất cả các phần tử trong một hàng (cách sử dụng rất phổ biến)print("a[1,:] = ", a[1,:], ",  a[1,:].shape =", a[1,:].shape, "a 1-D array")
# giống nhưprint("a[1]   = ", a[1],   ",  a[1].shape   =", a[1].shape, "a 1-D array")
```
