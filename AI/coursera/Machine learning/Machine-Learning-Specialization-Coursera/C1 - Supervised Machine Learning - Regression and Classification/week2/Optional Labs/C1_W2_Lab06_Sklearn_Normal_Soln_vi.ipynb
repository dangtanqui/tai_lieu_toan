# Linear Regression, giải pháp dạng đóng
Scikit-learn có [linear regression model](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html#sklearn.linear_model.LinearRegression) triển khai linear regression dạng đóng.

Hãy sử dụng dữ liệu từ các lab ban đầu - một ngôi nhà rộng 1000 feet vuông được bán với giá \$300,000 và một ngôi nhà rộng 2000 mét vuông được bán với giá \$500,000.

| Kích thước (1000 ft vuông) | Giá (1000 đô la) |
| ----------------| ------------------------ |
| 1 | 300 |
| 2 | 500 |


### Tải tập dữ liệu


```python
X_train = np.array([1.0, 2.0])   #features
y_train = np.array([300, 500])   #target value
```

### Tạo và lắp model
Code bên dưới thực hiện regression bằng scikit-learn. 
Bước đầu tiên tạo đối tượng regression.  
Bước thứ hai sử dụng một trong các phương thức được liên kết với đối tượng, `fit`. Việc này thực hiện regression, khớp parameters với dữ liệu input. Bộ công cụ mong đợi X matrix hai chiều.


```python
linear_model = LinearRegression()
# X phải là Matrix 2-Dlinear_model.fit(X_train.reshape(-1, 1), y_train) 
```

### Xem Parameters 
$\mathbf{w}$ và $\mathbf{b}$ parameters được gọi là 'hệ số' và 'chặn' trong scikit-learn.


```python
b = linear_model.intercept_
w = linear_model.coef_
print(f"w = {w:}, b = {b:0.2f}")
print(f"'manual' prediction: f_wb = wx+b : {1200*w + b}")
```

### Tạo Predictions

Việc gọi hàm `predict` sẽ tạo ra predictions.


```python
y_pred = linear_model.predict(X_train.reshape(-1, 1))

print("Prediction on training set:", y_pred)

X_test = np.array([[1200]])
print(f"Prediction for 1200 sqft house: ${linear_model.predict(X_test)[0]:0.2f}")
```

## Ví dụ thứ hai
Ví dụ thứ hai là từ lab trước đó có nhiều features. Các giá trị parameter cuối cùng và predictions rất gần với kết quả từ 'dài hạn' không được chuẩn hóa từ lab đó. Quá trình chạy không chuẩn hóa đó mất hàng giờ để tạo ra kết quả, trong khi điều này gần như xảy ra ngay lập tức. Giải pháp dạng đóng hoạt động tốt trên các tập dữ liệu nhỏ hơn như thế này nhưng có thể đòi hỏi tính toán cao trên các tập dữ liệu lớn hơn. 
>Giải pháp dạng đóng không yêu cầu normalization.


```python
# tải datasetX_train, y_train = load_house_data()
X_features = ['size(sqft)','bedrooms','floors','age']
```

```python
linear_model = LinearRegression()
linear_model.fit(X_train, y_train) 
```

```python
b = linear_model.intercept_
w = linear_model.coef_
print(f"w = {w:}, b = {b:0.2f}")
```

```python
print(f"Prediction on training set:\n {linear_model.predict(X_train)[:4]}" )
print(f"prediction using w,b:\n {(X_train @ w + b)[:4]}")
print(f"Target values \n {y_train[:4]}")

x_house = np.array([1200, 3,1, 40]).reshape(-1,4)
x_house_predict = linear_model.predict(x_house)[0]
print(f" predicted price of a house with 1200 sqft, 3 bedrooms, 1 floor, 40 years old = ${x_house_predict*1000:0.2f}")
```
