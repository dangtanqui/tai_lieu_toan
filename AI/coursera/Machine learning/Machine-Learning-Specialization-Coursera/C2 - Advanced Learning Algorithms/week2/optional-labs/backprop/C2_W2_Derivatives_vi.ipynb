## Định nghĩa không chính thức về đạo hàm


Định nghĩa chính thức về đạo hàm có thể hơi khó khăn với các giới hạn và giá trị 'bằng 0'. Ý tưởng thực sự đơn giản hơn nhiều. 

Đạo hàm của hàm mô tả cách đầu ra của hàm thay đổi khi có một thay đổi nhỏ ở biến đầu vào.

Hãy sử dụng hàm cost $J(w)$ làm ví dụ. cost $J$ là đầu ra và $w$ là biến đầu vào.  
Hãy đặt một 'thay đổi nhỏ' cho tên *epsilon* hoặc $\epsilon$. chúng ta sử dụng những chữ cái Hy Lạp này vì trong toán học truyền thống sử dụng *epsilon*($\epsilon$) hoặc *delta* ($\Delta$) để biểu thị một giá trị nhỏ. Bạn có thể coi nó đại diện cho 0,001 hoặc một số giá trị nhỏ khác.  

$$
\begin{equation}
\text{if } w \uparrow \epsilon \text{ causes }J(w) \uparrow \text{by }k \times \epsilon \text{ then}  \\
\frac{\partial J(w)}{\partial w} = k \tag{1}
\end{equation}
$$

Điều này chỉ cho biết nếu bạn thay đổi đầu vào của hàm $J(w)$ một chút và đầu ra thay đổi $k$ lần chút ít đó, thì đạo hàm của $J(w)$ sẽ bằng $k$.

Hãy thử điều này.  Hãy xét đạo hàm của hàm $J(w) = w^2$ tại điểm $w=3$ và $\epsilon = 0.001$


````python
J = (3)**2
J_epsilon = (3 + 0.001)**2
k = (J_epsilon - J)/0.001    # difference divided by epsilon
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k:0.6f} ")
````

chúng ta đã tăng giá trị đầu vào lên một chút (0,001), khiến đầu ra thay đổi từ 9 thành 9,006001, tăng gấp 6 lần mức tăng đầu vào. Tham khảo (1) ở trên, điều này cho biết rằng $k=6$, vì vậy $\frac{\partial J(w)}{\partial w} \approx 6$. Nếu bạn quen thuộc với phép tính, bạn biết đấy, được viết dưới dạng tượng trưng, ​​$\frac{\partial J(w)}{\partial w} = 2 w$. Với $w=3$ thì đây là 6. Phép tính của chúng ta ở trên không chính xác là 6 vì để chính xác thì $\epsilon$ cần phải là [infinitesimally small](https://www.dictionary.com/browse/infinitesimally) hoặc thực sự rất nhỏ. Đó là lý do tại sao chúng ta sử dụng ký hiệu $\approx$ hoặc ~= thay vì =. Hãy xem điều gì sẽ xảy ra nếu chúng ta làm cho $\epsilon$ nhỏ hơn.


````python
J = (3)**2
J_epsilon = (3 + 0.000000001)**2
k = (J_epsilon - J)/0.000000001
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k} ")
````

Giá trị tiến gần đến chính xác 6 khi chúng ta giảm kích thước $\epsilon$. Hãy thử giảm giá trị hơn nữa.


## Tìm đạo hàm tượng trưng
Trong backprop, việc biết đạo hàm của các hàm đơn giản ở bất kỳ giá trị đầu vào nào là rất hữu ích. Nói cách khác, chúng ta muốn biết đạo hàm 'ký hiệu' hơn là đạo hàm 'số học'. Một ví dụ về đạo hàm ký hiệu là $\frac{\partial J(w)}{\partial w} = 2 w$, đạo hàm của $J(w) = w^2$ ở trên.  Với đạo hàm ký hiệu, bạn có thể tìm giá trị của đạo hàm ở bất kỳ giá trị đầu vào nào $w$.  

Nếu bạn đã tham gia một khóa học giải tích, bạn đã quen thuộc với nhiều [differentiation rules](https://en.wikipedia.org/wiki/Differentiation_rules#Power_laws,_polynomials,_quotients,_and_reciprocals) mà các nhà toán học đã phát triển để giải đạo hàm cho một biểu thức. Chà, hóa ra quá trình này đã được tự động hóa bằng các chương trình phân biệt biểu tượng. Một ví dụ về điều này trong python là thư viện [SymPy](https://www.sympy.org/en/index.html). Chúng ta hãy xem làm thế nào để sử dụng cái này.


### $J = w^2$
Xác định các biến python và tên tượng trưng của chúng.


````python
J, w = symbols('J, w')
````

Xác định và in biểu thức. Lưu ý SymPy tạo ra chuỗi [latex](https://en.wikibooks.org/wiki/LaTeX/Mathematics) tạo ra một phương trình dễ đọc.


````python
J=w**2
J
````

Sử dụng SymPy's `diff` để phân biệt biểu thức của $J$ đối với $w$. Lưu ý kết quả phù hợp với ví dụ trước đó của chúng ta.


````python
dJ_dw = diff(J,w)
dJ_dw
````

Đánh giá đạo hàm tại một vài điểm bằng cách 'thay thế' các giá trị số cho các giá trị ký hiệu. Trong ví dụ đầu tiên, $w$ được thay thế bằng $2$.


````python
dJ_dw.subs([(w,2)])    # derivative at the point w = 2
````

````python
dJ_dw.subs([(w,3)])    # derivative at the point w = 3
````

````python
dJ_dw.subs([(w,-3)])    # derivative at the point w = -3
````

## $J = 2w$


````python
w, J = symbols('w, J')
````

````python
J = 2 * w
J
````

````python
dJ_dw = diff(J,w)
dJ_dw
````

````python
dJ_dw.subs([(w,-3)])    # derivative at the point w = -3
````

So sánh điều này với phép tính số học


````python
J = 2*3
J_epsilon = 2*(3 + 0.001)
k = (J_epsilon - J)/0.001
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k} ")
````

Đối với hàm $J=2w$, dễ dàng nhận thấy rằng bất kỳ thay đổi nào trong $w$ sẽ dẫn đến lượng thay đổi gấp 2 lần ở đầu ra $J$, bất kể giá trị bắt đầu của $w$. NumPy và kết quả số học của chúng ta xác nhận điều này.


## $J = w^3$


````python
J, w = symbols('J, w')
````

````python
J=w**3
J
````

````python
dJ_dw = diff(J,w)
dJ_dw
````

````python
dJ_dw.subs([(w,2)])   # derivative at the point w=2
````

So sánh điều này với phép tính số học


````python
J = (2)**3
J_epsilon = (2+0.001)**3
k = (J_epsilon - J)/0.001
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k} ")
````

## $J = \frac{1}{w}$


````python
J, w = symbols('J, w')
````

````python
J= 1/w
J
````

````python
dJ_dw = diff(J,w)
dJ_dw
````

````python
J = 1/2
J_epsilon = 1/(2+0.001)
k = (J_epsilon - J)/0.001
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k} ")
````

## $J = \frac{1}{w^2}$


````python
J, w = symbols('J, w')
````

Nếu bạn có thời gian, hãy thử lặp lại các bước trên với hàm $J = \frac{1}{w^2}$ và đánh giá tại w=4
